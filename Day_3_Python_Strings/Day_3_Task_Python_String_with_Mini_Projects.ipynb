1. **Complex String Manipulation with Loops**  
   - **Question**: Given a string of mixed data containing symbols and prices like `"AAPL100GOOG200MSFT300"`, write a script to extract and print each symbol with its corresponding price on separate lines.

```python
mix_str = "AAPL100GOOG200MSFT300"

current_symbol = ""
current_price = ""

for i in mix_str:
    if i.isalpha():
        if current_price != "":
            print(f"{current_symbol}: {current_price}")
            current_symbol = ""
            current_price = ""
        
        current_symbol += i
    
    elif i.isdigit():
        current_price += i

if current_symbol != "":
    print(f"{current_symbol}: {current_price}")
```

   - **Question**: Given a long string of security transaction details formatted as "Buy100AAPLSell200GOOG", where each segment starts with a transaction type followed by the amount and the ticker, print each transaction detail on a new line.

4. **Nested Loops with String Functions**
   - **Question**: Create a nested loop that reads a string of stock symbols separated by commas, then prints each symbol and its reverse.

```python
# Write your code here
stock_symbols = "AMZN,MFST,AAPL"
temp = ''
for i in stock_symbols: 
    if i != ',':
        temp += i
    if i == ',':
        reversed_temp = temp[::-1]
        print(temp,"->",reversed_temp)
        temp = ''

reversed_temp = temp[::-1] 
print(temp,"->",reversed_temp)

```

6. **Complex Data Extraction**
   - **Question**: Extract stock symbols and their respective share counts from a string like "Buy 50 AAPL and sell 30 GOOG", ensuring that only valid transactions (Buy/Sell) are processed.

```python
# Write your code here
str = "Buy 50 AAPL and sell 30 GOOG"
words = str.split()

for i in range(len(words)):
    if words[i] == "Buy" or words[i] == "sell":
        action = words[i]
        quantity = words[i+1]
        symbols = words[i+2]
        print(f"Action:{action},Quantity:{quantity},Symbols:{symbols}")

```

8. **String and Loop Integration**
   - **Question**: From a formatted string containing dates and prices like "20210101$10020210102$200", extract and print the date and price pairs.

```python
#Write your code here
formatted_str = "20210101$10020210102$200"
x = 12

for i in range(0,len(formatted_str)//x):
    data = formatted_str[x*i:x*(i+1)]
    split_data = data.split("$")
    
    print(f"Date: {split_data[0]}, Price: {split_data[1]}")
```

9. **String Parsing and Control Flow**
   - **Question**: Parse a string "High: 100, Low: 90, Open: 95, Close: 98" and print each value only if it is higher than 95.

```python
#Write your code here
s1 = "High: 100, Low: 90, Open: 95, Close: 98"
parts = s1.split(", ")

for i in parts:
    key,value = i.split(":")
    value = int(value)
   
    if value > 95:
        print(key,":",value)
```

### Project: Moving Average Crossover Trading Strategy (with list input)

#### Step-by-Step Plan

```python
# 1. Introduction
print("=== Moving Average Trading Strategy Simulator ===")
print("Strategy: Buy when Price > 3-Day MA, Sell otherwise.\n")

# 2. Reading Data
data = [
    ("2021-01-01", 100), ("2021-01-02", 102), ("2021-01-03", 104),
    ("2021-01-04", 101), ("2021-01-05", 105), ("2021-01-06", 107),
    ("2021-01-07", 110), ("2021-01-08", 108), ("2021-01-09", 106),
    ("2021-01-10", 104), ("2021-01-11", 103), ("2021-01-12", 102),
    ("2021-01-13", 101), ("2021-01-14", 105), ("2021-01-15", 108),
    ("2021-01-16", 110), ("2021-01-17", 112), ("2021-01-18", 115),
    ("2021-01-19", 117), ("2021-01-20", 120),
]

# Portfolio variables
balance = 1000  # Starting cash in $
shares_held = 0
window = 3

# 3, 4, & 5. Processing, Signaling, and Simulating
for i in range(window, len(data)):
    
    last_3_days = [data[i-1][1], data[i-2][1], data[i-3][1]]
    ma = sum(last_3_days) / window
    
    current_date = data[i][0]
    current_price = data[i][1]
    
    # 4. Generating Signals & 5. Simulating Trades
    print(f"Date: {current_date} | Price: {current_price} | MA: {ma:.2f}")
    
    if current_price > ma and balance >= current_price:
        # BUY Logic
        shares_to_buy = balance // current_price
        balance -= (shares_to_buy * current_price)
        shares_held += shares_to_buy
        print(f"--> SIGNAL: BUY {shares_to_buy} shares")
        
    elif current_price < ma and shares_held > 0:
        # SELL Logic
        balance += (shares_held * current_price)
        print(f"--> SIGNAL: SELL all shares at {current_price}")
        shares_held = 0
    
    print("-" * 40)

# 6. Printing Final Results
final_value = balance + (shares_held * data[-1][1])
profit = final_value - 1000

print("\n=== FINAL REPORT ===")
print(f"Final Portfolio Value: ${final_value:.2f}")
print(f"Total Profit/Loss: ${profit:.2f}")
```
